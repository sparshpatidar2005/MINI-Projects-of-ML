```python
import tensorflow as tf
from tensorflow.keras.applications import ResNet50
from tensorflow.keras import layers, models, optimizers, callbacks
from tensorflow.keras.utils import plot_model
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc, precision_recall_curve
from sklearn.utils import class_weight
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from datetime import datetime

# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Constants
BASE_DIR = '/content/drive/MyDrive/dataset'
MAGNIFICATIONS = ['40X', '100X', '200X', '400X']
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 15
TEST_SIZE = 0.15
VAL_SIZE = 0.15
SEED = 42

# Set random seeds
tf.random.set_seed(SEED)
np.random.seed(SEED)

# 1. Enhanced Data Pipeline with Proper Splitting
def create_data_pipeline(base_dir):
    # Collect all image paths with metadata
    data = []
    for root, dirs, files in os.walk(base_dir):
        current_mag = next((mag for mag in MAGNIFICATIONS if mag in root), None)
        if current_mag:
            for file in files:
                if file.endswith('.png') and not file.startswith('._'):
                    label = 0 if 'benign' in root.lower() else 1
                    data.append({
                        'path': os.path.join(root, file),
                        'label': label,
                        'magnification': current_mag
                    })

    # Create DataFrame
    df = pd.DataFrame(data)

    # Stratified split - first into train+val vs test
    stratify_col = df['label'].astype(str) + '_' + df['magnification']
    train_val_df, test_df = train_test_split(
        df,
        test_size=TEST_SIZE,
        stratify=stratify_col,
        random_state=SEED
    )

    # Second split - train vs val
    train_df, val_df = train_test_split(
        train_val_df,
        test_size=VAL_SIZE/(1-TEST_SIZE),
        stratify=train_val_df['label'].astype(str) + '_' + train_val_df['magnification'],
        random_state=SEED
    )

    # Calculate class weights
    class_weights = class_weight.compute_class_weight(
        'balanced',
        classes=np.unique(train_df['label']),
        y=train_df['label']
    )
    class_weights = dict(enumerate(class_weights))

    # Image preprocessing
    def preprocess_image(path, label):
        image = tf.io.read_file(path)
        image = tf.image.decode_png(image, channels=3)
        image = tf.image.resize(image, IMG_SIZE)
        image = tf.keras.applications.resnet50.preprocess_input(image)
        return image, label

    # Data augmentation
    def augment_image(image, label):
        image = tf.image.random_flip_left_right(image)
        image = tf.image.random_flip_up_down(image)
        image = tf.image.random_brightness(image, 0.1)
        image = tf.image.random_contrast(image, 0.9, 1.1)
        return image, label

    # Create TensorFlow datasets
    def df_to_dataset(df, shuffle=True, augment=False):
        ds = tf.data.Dataset.from_tensor_slices((df['path'].values, df['label'].values))
        if shuffle:
            ds = ds.shuffle(buffer_size=len(df), seed=SEED)
        ds = ds.map(preprocess_image, num_parallel_calls=tf.data.AUTOTUNE)
        if augment:
            ds = ds.map(augment_image, num_parallel_calls=tf.data.AUTOTUNE)
        ds = ds.batch(BATCH_SIZE)
        ds = ds.prefetch(tf.data.AUTOTUNE)
        return ds

    train_ds = df_to_dataset(train_df, shuffle=True, augment=True)
    val_ds = df_to_dataset(val_df, shuffle=False)
    test_ds = df_to_dataset(test_df, shuffle=False)

    return train_ds, val_ds, test_ds, test_df, class_weights

# Create datasets
train_dataset, val_dataset, test_dataset, test_df, class_weights = create_data_pipeline(BASE_DIR)

# 2. ResNet50 Fine-Tuning Model
def build_resnet50_model():
    base_model = ResNet50(
        weights='imagenet',
        include_top=False,
        input_shape=(224, 224, 3)
    )

    # Freeze first 100 layers
    for layer in base_model.layers[:100]:
        layer.trainable = False

    # Custom classification head
    inputs = layers.Input(shape=(224, 224, 3))
    x = base_model(inputs, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dense(256, activation='relu')(x)
    x = layers.Dropout(0.5)(x)
    outputs = layers.Dense(1, activation='sigmoid')(x)

    model = models.Model(inputs, outputs)

    # Compile with lower learning rate
    model.compile(
        optimizer=optimizers.Adam(learning_rate=1e-4),
        loss='binary_crossentropy',
        metrics=[
            'accuracy',
            tf.keras.metrics.Precision(name='precision'),
            tf.keras.metrics.Recall(name='recall'),
            tf.keras.metrics.AUC(name='auc'),
            tf.keras.metrics.AUC(name='prc', curve='PR')
        ]
    )

    return model

model = build_resnet50_model()
model.summary()

# 3. Training with Enhanced Callbacks
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

callbacks = [
    callbacks.EarlyStopping(
        monitor='val_prc',
        patience=5,
        mode='max',
        restore_best_weights=True,
        verbose=1
    ),
    callbacks.ModelCheckpoint(
        f'resnet50_best_{timestamp}.h5',
        monitor='val_prc',
        save_best_only=True,
        mode='max',
        verbose=1
    ),
    callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=3,
        min_lr=1e-6,
        verbose=1
    ),
    callbacks.TensorBoard(
        log_dir=f'logs/resnet50_{timestamp}',
        histogram_freq=1
    )
]

# Train model
history = model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=EPOCHS,
    callbacks=callbacks,
    class_weight=class_weights
)

# 4. Visualization Functions
def plot_training_history(history):
    plt.figure(figsize=(20, 12))

    metrics = ['accuracy', 'loss', 'auc', 'prc', 'precision', 'recall']
    titles = ['Accuracy', 'Loss', 'ROC AUC', 'PR AUC', 'Precision', 'Recall']

    for i, metric in enumerate(metrics, 1):
        plt.subplot(2, 3, i)
        plt.plot(history.history[metric], label='Train')
        plt.plot(history.history[f'val_{metric}'], label='Validation')
        plt.title(titles[i-1])
        plt.ylabel(titles[i-1])
        plt.xlabel('Epoch')
        plt.legend()

    plt.tight_layout()
    plt.show()

def evaluate_model(model, dataset, df):
    y_probs = model.predict(dataset).flatten()
    y_pred = (y_probs > 0.5).astype(int)
    y_true = df['label'].values

    # Classification report
    print("\nClassification Report:")
    print(classification_report(y_true, y_pred, target_names=['Benign', 'Malignant']))

    # Confusion matrix
    plt.figure(figsize=(6,6))
    cm = confusion_matrix(y_true, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=['Benign', 'Malignant'],
                yticklabels=['Benign', 'Malignant'])
    plt.ylabel('Actual')
    plt.xlabel('Predicted')
    plt.title('Confusion Matrix')
    plt.show()

    # ROC Curve
    fpr, tpr, _ = roc_curve(y_true, y_probs)
    roc_auc = auc(fpr, tpr)
    plt.figure(figsize=(8,6))
    plt.plot(fpr, tpr, color='darkorange', label=f'ROC (AUC = {roc_auc:.2f})')
    plt.plot([0, 1], [0, 1], color='navy', linestyle='--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title('ROC Curve')
    plt.legend()
    plt.show()

    # Precision-Recall Curve
    precision, recall, _ = precision_recall_curve(y_true, y_probs)
    pr_auc = auc(recall, precision)
    plt.figure(figsize=(8,6))
    plt.plot(recall, precision, color='blue', label=f'PR (AUC = {pr_auc:.2f})')
    plt.xlabel('Recall')
    plt.ylabel('Precision')
    plt.title('Precision-Recall Curve')
    plt.legend()
    plt.show()

    return {
        'y_true': y_true,
        'y_pred': y_pred,
        'y_probs': y_probs,
        'classification_report': classification_report(y_true, y_pred, output_dict=True),
        'roc_auc': roc_auc,
        'pr_auc': pr_auc
    }

def magnification_analysis(df, y_true, y_pred, y_probs):
    results = {}
    plt.figure(figsize=(15, 10))

    for i, mag in enumerate(MAGNIFICATIONS, 1):
        mask = df['magnification'] == mag
        if sum(mask) == 0:
            continue

        y_true_mag = y_true[mask]
        y_pred_mag = y_pred[mask]
        y_probs_mag = y_probs[mask]

        # Classification report
        report = classification_report(y_true_mag, y_pred_mag, output_dict=True)

        # Confusion matrix
        cm = confusion_matrix(y_true_mag, y_pred_mag)

        # ROC and PR AUC
        fpr, tpr, _ = roc_curve(y_true_mag, y_probs_mag)
        roc_auc = auc(fpr, tpr)
        precision, recall, _ = precision_recall_curve(y_true_mag, y_probs_mag)
        pr_auc = auc(recall, precision)

        results[mag] = {
            'accuracy': report['accuracy'],
            'precision': report['weighted avg']['precision'],
            'recall': report['weighted avg']['recall'],
            'f1': report['weighted avg']['f1-score'],
            'roc_auc': roc_auc,
            'pr_auc': pr_auc
        }

        # Plot confusion matrix
        plt.subplot(2, 2, i)
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                   xticklabels=['Benign', 'Malignant'],
                   yticklabels=['Benign', 'Malignant'])
        plt.title(f'{mag} (Acc: {report["accuracy"]:.2f}, AUC: {roc_auc:.2f})')
        plt.ylabel('Actual')
        plt.xlabel('Predicted')

    plt.tight_layout()
    plt.show()

    # Create comparison table
    metrics_df = pd.DataFrame(results).T
    print("\nMagnification-Level Performance:")
    display(metrics_df.style.background_gradient(cmap='Blues'))

    return results

# 5. Run Evaluations
print("\n=== Training History ===")
plot_training_history(history)

print("\n=== Test Set Evaluation ===")
test_results = evaluate_model(model, test_dataset, test_df)

print("\n=== Magnification Analysis ===")
mag_results = magnification_analysis(test_df, test_results['y_true'],
                                   test_results['y_pred'], test_results['y_probs'])

# 6. Final Summary
print("\n=== FINAL MODEL SUMMARY ===")
print(f"Trained for {len(history.history['loss'])}/{EPOCHS} epochs")
print(f"Best Validation PR AUC: {max(history.history['val_prc']):.4f}")
print("\nTest Set Metrics:")
print(f"Accuracy: {test_results['classification_report']['accuracy']:.4f}")
print(f"Precision: {test_results['classification_report']['weighted avg']['precision']:.4f}")
print(f"Recall: {test_results['classification_report']['weighted avg']['recall']:.4f}")
print(f"F1-Score: {test_results['classification_report']['weighted avg']['f1-score']:.4f}")
print(f"ROC AUC: {test_results['roc_auc']:.4f}")
print(f"PR AUC: {test_results['pr_auc']:.4f}")
```
