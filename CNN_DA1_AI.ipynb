```python
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers, callbacks
from tensorflow.keras.utils import plot_model
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc, precision_recall_curve
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from datetime import datetime

# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Constants
BASE_DIR = '/content/drive/MyDrive/dataset'
MAGNIFICATIONS = ['40X', '100X', '200X', '400X']
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 15
TEST_SIZE = 0.15
VAL_SIZE = 0.15
SEED = 42

# Set random seeds
tf.random.set_seed(SEED)
np.random.seed(SEED)

# 1. Enhanced Data Pipeline
def create_data_pipeline(base_dir):
    # Collect all image paths with metadata
    data = []
    for root, dirs, files in os.walk(base_dir):
        current_mag = next((mag for mag in MAGNIFICATIONS if mag in root), None)
        if current_mag:
            for file in files:
                if file.endswith('.png') and not file.startswith('._'):
                    label = 0 if 'benign' in root.lower() else 1
                    data.append({
                        'path': os.path.join(root, file),
                        'label': label,
                        'magnification': current_mag
                    })

    # Create DataFrame
    df = pd.DataFrame(data)

    # Stratified split into train/val/test
    stratify_col = df['label'].astype(str) + '_' + df['magnification']
    train_val_df, test_df = train_test_split(
        df, test_size=TEST_SIZE, stratify=stratify_col, random_state=SEED
    )
    train_df, val_df = train_test_split(
        train_val_df,
        test_size=VAL_SIZE/(1-TEST_SIZE),
        stratify=train_val_df['label'].astype(str) + '_' + train_val_df['magnification'],
        random_state=SEED
    )

    # Print dataset sizes
    print("\nDataset Split Sizes:")
    print(f"Training: {len(train_df)} samples")
    print(f"Validation: {len(val_df)} samples")
    print(f"Testing: {len(test_df)} samples")

    # Print class distribution
    print("\nClass Distribution:")
    print(train_df['label'].value_counts())

    # Image preprocessing
    def preprocess_image(path, label):
        image = tf.io.read_file(path)
        image = tf.image.decode_png(image, channels=3)
        image = tf.image.resize(image, IMG_SIZE)
        image = tf.keras.applications.resnet50.preprocess_input(image)
        return image, label

    # Data augmentation
    def augment_image(image, label):
        image = tf.image.random_flip_left_right(image)
        image = tf.image.random_flip_up_down(image)
        image = tf.image.random_brightness(image, 0.1)
        image = tf.image.random_contrast(image, 0.9, 1.1)
        return image, label

    # Create datasets
    def df_to_dataset(df, shuffle=True, augment=False):
        ds = tf.data.Dataset.from_tensor_slices((df['path'].values, df['label'].values))
        if shuffle:
            ds = ds.shuffle(buffer_size=len(df), seed=SEED)
        ds = ds.map(preprocess_image, num_parallel_calls=tf.data.AUTOTUNE)
        if augment:
            ds = ds.map(augment_image, num_parallel_calls=tf.data.AUTOTUNE)
        ds = ds.batch(BATCH_SIZE)
        ds = ds.prefetch(tf.data.AUTOTUNE)
        return ds

    train_ds = df_to_dataset(train_df, shuffle=True, augment=True)
    val_ds = df_to_dataset(val_df, shuffle=False)
    test_ds = df_to_dataset(test_df, shuffle=False)

    return train_ds, val_ds, test_ds, train_df, val_df, test_df

# Create datasets
train_dataset, val_dataset, test_dataset, train_df, val_df, test_df = create_data_pipeline(BASE_DIR)

# 2. 7-Layer CNN Model Architecture
def build_7layer_cnn(input_shape=(224, 224, 3)):
    inputs = layers.Input(shape=input_shape)

    # Block 1
    x = layers.Conv2D(32, (3,3), activation='relu', padding='same')(inputs)
    x = layers.BatchNormalization()(x)
    x = layers.Conv2D(32, (3,3), activation='relu', padding='same')(x)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling2D((2,2))(x)
    x = layers.Dropout(0.2)(x)

    # Block 2
    x = layers.Conv2D(64, (3,3), activation='relu', padding='same')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Conv2D(64, (3,3), activation='relu', padding='same')(x)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling2D((2,2))(x)
    x = layers.Dropout(0.3)(x)

    # Block 3
    x = layers.Conv2D(128, (3,3), activation='relu', padding='same')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Conv2D(128, (3,3), activation='relu', padding='same')(x)
    x = layers.BatchNormalization()(x)
    x = layers.MaxPooling2D((2,2))(x)
    x = layers.Dropout(0.4)(x)

    # Block 4 (Additional layers to make 7 conv layers)
    x = layers.Conv2D(256, (3,3), activation='relu', padding='same')(x)
    x = layers.BatchNormalization()(x)
    x = layers.Conv2D(256, (3,3), activation='relu', padding='same')(x)
    x = layers.BatchNormalization()(x)
    x = layers.GlobalAveragePooling2D()(x)

    # Classification head
    x = layers.Dense(512, activation='relu')(x)
    x = layers.Dropout(0.5)(x)
    outputs = layers.Dense(1, activation='sigmoid')(x)

    model = models.Model(inputs, outputs)

    model.compile(
        optimizer=optimizers.Adam(learning_rate=1e-4),
        loss='binary_crossentropy',
        metrics=[
            'accuracy',
            tf.keras.metrics.Precision(name='precision'),
            tf.keras.metrics.Recall(name='recall'),
            tf.keras.metrics.AUC(name='auc'),
            tf.keras.metrics.AUC(name='prc', curve='PR')
        ]
    )

    return model

model = build_7layer_cnn()
model.summary()
plot_model(model, to_file='7layer_cnn_architecture.png', show_shapes=True)

# 3. Training Setup
callbacks = [
    callbacks.EarlyStopping(
        monitor='val_prc',
        patience=5,
        mode='max',
        restore_best_weights=True,
        verbose=1
    ),
    callbacks.ModelCheckpoint(
        'best_7layer_model.h5',
        monitor='val_prc',
        save_best_only=True,
        verbose=1
    ),
    callbacks.TensorBoard(
        log_dir='logs/7layer_cnn'
    ),
    callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.5,
        patience=3,
        min_lr=1e-6,
        verbose=1
    )
]

# 4. Training
print("\nStarting training for 7-layer CNN...")
history = model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=EPOCHS,
    callbacks=callbacks,
    verbose=1
)

# 5. Visualization and Evaluation Functions
def plot_training_metrics(history):
    plt.figure(figsize=(20, 12))

    metrics = ['loss', 'accuracy', 'precision', 'recall', 'auc', 'prc']
    titles = ['Loss', 'Accuracy', 'Precision', 'Recall', 'ROC AUC', 'PR AUC']

    for i, metric in enumerate(metrics, 1):
        plt.subplot(2, 3, i)
        plt.plot(history.history[metric], label='Train')
        plt.plot(history.history[f'val_{metric}'], label='Validation')
        plt.title(titles[i-1])
        plt.xlabel('Epoch')
        plt.legend()

    plt.tight_layout()
    plt.show()

def evaluate_dataset(model, dataset, df, dataset_name):
    y_probs = model.predict(dataset).flatten()
    y_pred = (y_probs > 0.5).astype(int)
    y_true = df['label'].values

    # Classification report
    print(f"\n{classification_report(y_true, y_pred, target_names=['Benign', 'Malignant'])}")

    # Confusion matrix
    plt.figure(figsize=(6,6))
    cm = confusion_matrix(y_true, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=['Benign', 'Malignant'],
                yticklabels=['Benign', 'Malignant'])
    plt.title(f'{dataset_name} Confusion Matrix')
    plt.show()

    # ROC curve
    fpr, tpr, _ = roc_curve(y_true, y_probs)
    roc_auc = auc(fpr, tpr)
    plt.figure(figsize=(8,6))
    plt.plot(fpr, tpr, label=f'ROC Curve (AUC = {roc_auc:.2f})')
    plt.plot([0, 1], [0, 1], 'k--')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.title(f'{dataset_name} ROC Curve')
    plt.legend()
    plt.show()

    # Precision-Recall curve
    precision, recall, _ = precision_recall_curve(y_true, y_probs)
    pr_auc = auc(recall, precision)
    plt.figure(figsize=(8,6))
    plt.plot(recall, precision, label=f'PR Curve (AUC = {pr_auc:.2f})')
    plt.xlabel('Recall')
    plt.ylabel('Precision')
    plt.title(f'{dataset_name} Precision-Recall Curve')
    plt.legend()
    plt.show()

    return {
        'y_true': y_true,
        'y_pred': y_pred,
        'y_probs': y_probs,
        'roc_auc': roc_auc,
        'pr_auc': pr_auc
    }

def evaluate_magnification_levels(model, df, dataset_name):
    results = {}
    plt.figure(figsize=(15, 10))

    for i, mag in enumerate(MAGNIFICATIONS, 1):
        mag_df = df[df['magnification'] == mag]
        if len(mag_df) == 0:
            continue

        # Create dataset for this magnification
        def preprocess_image(path, label):
            image = tf.io.read_file(path)
            image = tf.image.decode_png(image, channels=3)
            image = tf.image.resize(image, IMG_SIZE)
            image = tf.keras.applications.resnet50.preprocess_input(image)
            return image, label

        mag_ds = tf.data.Dataset.from_tensor_slices((mag_df['path'].values, mag_df['label'].values))
        mag_ds = mag_ds.map(preprocess_image).batch(BATCH_SIZE)

        # Evaluate
        y_probs = model.predict(mag_ds).flatten()
        y_pred = (y_probs > 0.5).astype(int)
        y_true = mag_df['label'].values

        # Metrics
        report = classification_report(y_true, y_pred, output_dict=True)
        fpr, tpr, _ = roc_curve(y_true, y_probs)
        roc_auc = auc(fpr, tpr)
        precision, recall, _ = precision_recall_curve(y_true, y_probs)
        pr_auc = auc(recall, precision)

        results[mag] = {
            'accuracy': report['accuracy'],
            'precision': report['weighted avg']['precision'],
            'recall': report['weighted avg']['recall'],
            'f1': report['weighted avg']['f1-score'],
            'roc_auc': roc_auc,
            'pr_auc': pr_auc
        }

        # Confusion matrix per magnification
        plt.subplot(2, 2, i)
        cm = confusion_matrix(y_true, y_pred)
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                   xticklabels=['Benign', 'Malignant'],
                   yticklabels=['Benign', 'Malignant'])
        plt.title(f'{mag} (Acc: {report["accuracy"]:.2f}, AUC: {roc_auc:.2f})')

    plt.suptitle(f'Magnification-Level Performance ({dataset_name})')
    plt.tight_layout()
    plt.show()

    # Create metrics table
    metrics_df = pd.DataFrame(results).T
    print(f"\nMagnification-Level Metrics ({dataset_name}):")
    display(metrics_df.style.background_gradient(cmap='Blues'))

    return results

# 6. Comprehensive Evaluation
print("\nTraining Metrics:")
plot_training_metrics(history)

print("\nValidation Set Evaluation:")
val_results = evaluate_dataset(model, val_dataset, val_df, "Validation")
val_mag_results = evaluate_magnification_levels(model, val_df, "Validation")

print("\nTest Set Evaluation:")
test_results = evaluate_dataset(model, test_dataset, test_df, "Test")
test_mag_results = evaluate_magnification_levels(model, test_df, "Test")

# 7. Final Summary
print("\n=== FINAL 7-LAYER CNN RESULTS ===")
print(f"Best Validation PR AUC: {max(history.history['val_prc']):.4f}")
print(f"Test ROC AUC: {test_results['roc_auc']:.4f}")
print(f"Test PR AUC: {test_results['pr_auc']:.4f}")

print("\nTest Set Magnification Performance:")
for mag, metrics in test_mag_results.items():
    print(f"\n{mag}:")
    print(f"Accuracy: {metrics['accuracy']:.4f}")
    print(f"Precision: {metrics['precision']:.4f}")
    print(f"Recall: {metrics['recall']:.4f}")
    print(f"F1-Score: {metrics['f1']:.4f}")
    print(f"ROC AUC: {metrics['roc_auc']:.4f}")
    print(f"PR AUC: {metrics['pr_auc']:.4f}")
```
